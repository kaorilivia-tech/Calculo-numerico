```python
import numpy as np


#ALGORITMO: ELIMINAÇÃO DE GAUSS (SEM PIVOTEAMENTO)

def eliminacao_gauss_sem_pivoteamento(A, b):
    """
    Resolve o sistema linear Ax = b utilizando o Método da Eliminação de Gauss
    sem pivoteamento com substituição regressiva.

    Parâmetros:
      A : array_like (matriz de coeficientes n x n)
      b : array_like (vetor de termos independentes n)

    Retorna:
      x : ndarray (vetor solução n)
    """
    #Converter entradas para float para evitar divisões inteiras

    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    #1. Construir a Matriz Aumentada [A|b]

    Aumentada = np.hstack([A, b.reshape(-1, 1)])

    #2. Etapa de Eliminação (Triangularização Superior)

    for i in range(n):
        pivo = Aumentada[i, i]
        if np.isclose(pivo, 0):
            raise ValueError(f"Pivô nulo detectado na posição ({i},{i}). O algoritmo sem pivoteamento falhou.")

        for j in range(i + 1, n):
            fator = Aumentada[j, i] / pivo
            Aumentada[j, i:] -= fator * Aumentada[i, i:]

    #3. Etapa de Substituição Regressiva
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.sum(Aumentada[i, i + 1:n] * x[i + 1:n])
        x[i] = (Aumentada[i, n] - soma) / Aumentada[i, i]

    return x




#PARTE 3 — EXERCÍCIO 4: TESTE DO ALGORITMO E COMPARAÇÃO COM NUMPY

# Definição do sistema do Exercício 2:
#  2x +  y -  z =   8
# -3x -  y + 2z = -11
# -2x +  y + 2z =  -3
A_ex2 = np.array([[2, 1, -1],
                  [-3, -1, 2],
                  [-2, 1, 2]])

b_ex2 = np.array([8, -11, -3])

#Resolução pelo algoritmo customizado
x_gauss = eliminacao_gauss_sem_pivoteamento(A_ex2, b_ex2)

#Resolução de referência via biblioteca NumPy
x_numpy = np.linalg.solve(A_ex2, b_ex2)

print("=" * 65)
print(" PARTE 3 — EXERCÍCIO 4: TESTE DO ALGORITMO")
print("=" * 65)
print(f"Solução calculada (Gauss Customizado): {x_gauss}")
print(f"Solução oficial (numpy.linalg.solve):  {x_numpy}")
print(f"Diferença absoluta:                     {np.abs(x_gauss - x_numpy)}")
print("\n[Interpretação Ex. 4]:")
print("A solução obtida pelo algoritmo [2., 3., -1.] é idêntica à fornecida")
print("pelo numpy.linalg.solve e aos cálculos manuais da Parte 1, validando o código.")
print("=" * 65)



#PARTE 4 — EXERCÍCIO 5: MODELO DE BALANÇO TÉRMICO

#Matriz e vetor do modelo térmico:
# [ 4 -1  0 ] [ T1 ]   [ 15 ]
# [-1  4 -1 ] [ T2 ] = [ 10 ]
# [ 0 -1  3 ] [ T3 ]   [ 10 ]
A_termico = np.array([[4, -1, 0],
                      [-1, 4, -1],
                      [0, -1, 3]])

b_termico = np.array([15, 10, 10])

#Resolução do sistema térmico
T_solucao = eliminacao_gauss_sem_pivoteamento(A_termico, b_termico)

print("\n" + "=" * 65)
print(" PARTE 4 — EXERCÍCIO 5: APLICAÇÃO (BALANÇO TÉRMICO)")
print("=" * 65)
print(f"Temperatura T1: {T_solucao[0]:.2f} °C")
print(f"Temperatura T2: {T_solucao[1]:.2f} °C")
print(f"Temperatura T3: {T_solucao[2]:.2f} °C")
print("\n[Interpretação Ex. 5]:")
print("O sistema atinge o equilíbrio quando todas as temperaturas internas se equalizam")
print("em 5 °C (T1 = 5, T2 = 5, T3 = 5). Fisicamente, isso ocorre porque as trocas de calor")
print("entre nós vizinhos e dissipações compensam exatamente a energia fornecida no sistema.")
print("=" * 65)
```
