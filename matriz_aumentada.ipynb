```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

M_0 = np.array([
    [ 2.0,  1.0, -1.0,   8.0],
    [-3.0, -1.0,  2.0, -11.0],
    [-2.0,  1.0,  2.0,  -3.0]
])

M_1 = np.array([
    [2.0, 1.0, -1.0, 8.0],
    [0.0, 0.5,  0.5, 1.0],
    [0.0, 2.0,  1.0, 5.0]
])

M_2 = np.array([
    [2.0, 1.0, -1.0, 8.0],
    [0.0, 0.5,  0.5, 1.0],
    [0.0, 0.0, -1.0, 1.0]
])

estados = [
    {
        "matriz": M_0,
        "titulo": "Matriz Aumentada Inicial [A | b]",
        "desc": "Pivô a11 = 2.0\nPreparando para eliminar abaixo da Coluna 1...",
        "pivo": (0, 0)
    },
    {
        "matriz": M_1,
        "titulo": "Passo 1: Eliminação Abaixo do 1º Pivô (a11 = 2)",
        "desc": "Multiplicadores: m21 = -1.5,  m31 = -1.0\nR2 ← R2 - (-1.5)R1\nR3 ← R3 - (-1.0)R1",
        "pivo": (1, 1)
    },
    {
        "matriz": M_2,
        "titulo": "Passo 2: Eliminação Abaixo do 2º Pivô (a22 = 0.5)",
        "desc": "Multiplicador: m32 = 4.0\nR3 ← R3 - (4.0)R2\nForma Triangular Superior Concluída!",
        "pivo": (2, 2)
    }
]

fig, ax = plt.subplots(figsize=(7, 4.5), dpi=100)
plt.close(fig)

def render_frame(frame):
    ax.clear()
    ax.set_xlim(0, 5)
    ax.set_ylim(0, 5)
    ax.axis('off')

    data = estados[frame]
    M = data["matriz"]
    pr, pc = data["pivo"]

    ax.text(2.5, 4.4, data["titulo"], fontsize=12, fontweight='bold', ha='center', color='#1a252f')

    headers = ["x", "y", "z", "|", "b"]
    xs = [1.0, 1.8, 2.6, 3.2, 3.8]
    for x, h in zip(xs, headers):
        ax.text(x, 3.6, h, fontsize=12, fontweight='bold', ha='center', color='#2c3e50')

    ax.plot([0.6, 4.2], [3.3, 3.3], color='#7f8c8d', lw=2)

    ys = [2.7, 2.0, 1.3]
    for r in range(3):
        y = ys[r]
        row_vals = [M[r, 0], M[r, 1], M[r, 2], "|", M[r, 3]]

        for c_idx, val in enumerate(row_vals):
            x = xs[c_idx]

            if c_idx < 3 and r == pr and c_idx == pc:
                rect = plt.Rectangle((x - 0.28, y - 0.25), 0.56, 0.5, facecolor='#ffeaa7', edgecolor='#fdcb6e', zorder=1)
                ax.add_patch(rect)

            txt = "|" if val == "|" else f"{val:g}"
            ax.text(x, y, txt, fontsize=12, ha='center', va='center', zorder=2,
                    fontweight='bold' if (c_idx < 3 and r == pr and c_idx == pc) else 'normal')

    ax.text(2.5, 0.4, data["desc"], ha='center', va='center', fontsize=10,
            bbox=dict(boxstyle='round,pad=0.5', facecolor='#dfe6e9', edgecolor='#b2bec3'))

ani = animation.FuncAnimation(fig, render_frame, frames=len(estados), interval=3000, repeat=True)

HTML(ani.to_jshtml())
```
